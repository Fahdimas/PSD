# Pemantauan Kualitas Udara (Polutan Atmosfer) Wilayah Bangkalan dan Sekitarnya

## 1. Business Understanding

### 1.1 Latar Belakang
Pemantauan kualitas udara merupakan langkah krusial untuk menjaga kesehatan lingkungan berkelanjutan. Wilayah Bangkalan, Madura, dan sekitarnya memiliki dinamika aktivitas masyarakat, mobilitas kendaraan (terutama di area penghubung Jembatan Suramadu), serta kondisi geografis pesisir yang dapat memengaruhi pergerakan polutan atmosfer. Saat ini, masih diperlukan visualisasi data historis yang mudah dipahami terkait tingkat polusi di area ini, khususnya untuk memantau tren tahunan.

### 1.2 Tujuan (Business Objectives)
1. Mengekstraksi data polutan udara secara spesifik pada batas wilayah koordinat Bangkalan dan sekitarnya menggunakan layanan API dari satelit Copernicus.
2. Mentransformasi format data geospasial mentah (NetCDF) menjadi struktur data tabular (CSV) untuk memudahkan pemrosesan komputasi.
3. Menganalisis fluktuasi harian kadar polutan utama selama periode satu tahun terakhir (Agustus 2025 - Agustus 2026) melalui visualisasi grafik *time series*.

### 1.3 Manfaat (Business Impacts)
* **Bagi Masyarakat:** Memberikan *insight* visual mengenai waktu atau bulan-bulan tertentu di mana polusi mencapai puncaknya, sebagai referensi kesehatan lingkungan.
* **Bagi Akademisi & Stakeholder:** Menjadi landasan data historis awal untuk analisis lebih lanjut terkait dampak lingkungan di wilayah pesisir Jawa Timur dan Madura.

## 2. Data Understanding

Data yang diolah dalam analisis ini ditarik dari dataset `cams-global-reanalysis-eac4` (Copernicus Atmosphere Monitoring Service). Pengambilan data dibatasi menggunakan *bounding box* koordinat yang meliputi perairan Madura hingga pesisir utara Surabaya.

Berdasarkan instruksi analisis, seluruh parameter polutan gas dan partikel utama diekstrak untuk membandingkan tingkat dominansinya di udara. Berikut adalah rincian polutan yang dianalisis:

* **CO (Karbon Monoksida):** Gas beracun tak berwarna yang bersumber dari pembakaran tidak sempurna, sering kali menjadi indikator emisi asap kendaraan bermotor.
* **NO2 (Nitrogen Dioksida):** Gas berbau tajam penyebab iritasi pernapasan, bersumber dari emisi kendaraan bermesin diesel dan aktivitas industri.
* **SO2 (Sulfur Dioksida):** Indikator polusi dari fasilitas pabrik atau pembakaran batubara, serta pemicu terjadinya hujan asam.
* **O3 (Ozon Permukaan Tanah):** Polutan sekunder yang terbentuk dari reaksi fotokimia polutan lain akibat paparan panas sinar matahari.
* **PM10 & PM2.5 (Particulate Matter):** Partikel padat mikroskopis (debu, asap, polusi partikel) yang melayang di udara. PM2.5 memiliki ukuran paling kecil (kurang dari 2.5 mikrometer) sehingga berisiko tertinggi karena dapat menembus sistem peredaran darah manusia.
