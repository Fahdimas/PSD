```python
import pandas as pd
import numpy as np
import tensorflow as tf

# 1. Muat Data yang sudah dibuat sebelumnya
df = pd.read_csv('data_timeseries_3kolom.csv')

# Mengambil 2 kolom numerik
features = df[['Penjualan', 'Pengunjung']].values

# Normalisasi Min-Max
features_min = features.min(axis=0)
features_max = features.max(axis=0)
features_normalized = (features - features_min) / (features_max - features_min)

# 2. Windowing (Jendela Waktu)
window_size = 3
dataset = tf.keras.utils.timeseries_dataset_from_array(
    data=features_normalized,
    targets=None,
    sequence_length=window_size,
    sequence_stride=1,
    batch_size=4
)

# 3. Model Ekstraktor Fitur
feature_extractor = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(window_size, 2)), 
    
    # LSTM diperbesar agar bisa menangkap pola lebih kompleks dari 2 kolom
    tf.keras.layers.LSTM(32, activation='tanh', return_sequences=False),
    
    # DI SINI KUNCINYA: Kita set output menjadi tepat 14 fitur
    tf.keras.layers.Dense(14, activation='relu', name='fitur_ekstraksi_14')
])

feature_extractor.summary()

# 4. Implementasi
print("\n--- HASIL EKSTRAKSI 14 FITUR ---")
for batch in dataset.take(1):
    extracted_features = feature_extractor.predict(batch)
    
    print(f"Bentuk Input (Batch): {batch.shape}")
    print(f"Bentuk Output: {extracted_features.shape} -> (Batch Size, 14 Fitur)")
    
    print("\nContoh 14 Fitur dari satu potongan jendela waktu:")
    print(extracted_features[0]) # Menampilkan 14 fitur untuk data pertama di batch
```
